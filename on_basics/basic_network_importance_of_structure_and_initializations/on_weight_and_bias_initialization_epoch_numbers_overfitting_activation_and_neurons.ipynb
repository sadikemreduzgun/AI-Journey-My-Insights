```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# 1. Generate training data
# Let's use numbers in a range, e.g. [-10, 10]
X = np.linspace(-10, 10, 1000).reshape(-1, 1)  # a vector
y = X.copy()  # because f(x) = x

# 2. Define the model
model = Sequential([
    Dense(1, input_shape=(1,), activation=None)  # one neuron, no activation
])

model.compile(optimizer='adam', loss='mse')

# 3. Train
model.fit(X, y, epochs=10, verbose=0)

# 4. Test
test_values = np.array([[-5], [0], [7], [10]])
predictions = model.predict(test_values)
print("Input:", test_values.flatten())
print("Predicted:", predictions.flatten())
```

# Wow learned something, this approach sucks as it can be seen. I asked to GPT and here is answer:

❌ Problem: Only 1 neuron + ReLU → loses negative values

Your input range is from -10 to 10.
ReLU does:
* relu(x) = max(0, x)

So for all negative inputs, the first layer outputs 0.
After training, the dense weights adjust based mostly on positive inputs.
End result: the model outputs nearly the same constant for everything.

1. Zero initialization for tiny networks + few epochs → network barely learns.

2. Randomness in Adam updates introduces tiny variations even across runs.

3. Network capacity is too small to represent f(x)=x with zero initialization in 10 epochs.

    If you want stable, perfect predictions, you should:
* Initialize weights closer to 1 (identity)
* Use more epochs
* Optionally, set seeds for reproducibility

### Why overfitting is hard to face in this case:
* Model: Dense(1) → Dense(1) → 2 neurons total
* Data: X has 1000 points from -10 to 10
* Target: y = X → perfect linear function

Observations:
* Network is tiny, only 2 weights + 2 biases → almost no capacity to overfit
* Data is dense and clean, no noise → overfitting isn’t an issue
* Training longer just allows weights to converge exactly to the solution

# let's start 2 hidden layer neurons with -0.5 and 2 with 0.5 and see result

```python
def run_model_custom_init(times):
    for i in range(times):
        model = Sequential([
            Dense(4, input_shape=(1,), activation='relu'),  # 4 hidden neurons
            Dense(1, activation=None)
        ])

        # Custom initialization for hidden layer
        # 2 neurons = 0.5, 2 neurons = -0.5
        # Dense weight shape: (input_dim, units) = (1, 4)
        # Dense bias shape: (units,) = (4,)
        hidden_weights = np.array([[0.5, 0.5, -0.5, -0.5]])  # shape (1,4)
        hidden_biases = np.zeros(4)
        model.layers[0].set_weights([hidden_weights, hidden_biases])

        # Output layer initialization (can stay default or custom)
        output_weights = np.array([[-0.5], [-0.5], [-0.5], [0.5]])  # shape (4,1)
        output_biases = np.zeros(1)
        model.layers[1].set_weights([output_weights, output_biases])

        model.compile(optimizer='adam', loss='mse')

        # Train
        model.fit(X, y, epochs=30, verbose=0)

        # Test
        test_values = np.array([[-5], [0], [7], [10]])
        predictions = model.predict(test_values)
        print("Input:", test_values.flatten())
        print("Predicted:", predictions.flatten())
        print("-" * 50)

run_model_custom_init(3)
```

```python
from tensorflow.keras.initializers import RandomUniform

# random range was around 1.095 but here -1 to 1 will be used for ease

def run_model_random_init(times):
    for i in range(times):
        model = Sequential([
            Dense(4, input_shape=(1,), activation='relu',
                  kernel_initializer=RandomUniform(minval=-1, maxval=1),
                  bias_initializer=RandomUniform(minval=-0.1, maxval=0.1)),
            Dense(1, activation=None,
                  kernel_initializer=RandomUniform(minval=-0.5, maxval=0.5),
                  bias_initializer=RandomUniform(minval=-0.1, maxval=0.1))
        ])

        # Compile
        model.compile(optimizer='adam', loss='mse')

        # Train
        model.fit(X, y, epochs=30, verbose=0)

        # Test
        test_values = np.array([[-5], [0], [7], [10]])
        predictions = model.predict(test_values)
        print("Run", i+1)
        print("Input:", test_values.flatten())
        print("Predicted:", predictions.flatten())
        print("-" * 50)

run_model_random_init(3)
```

# well well well, you have been deceived 😎
### maybe you thought: Input data goes sequentially from -10 → 10
### ReLU kills negative pre-activations:
* Neurons with positive weights → pre-activation = w*x
* For x < 0 → w*x < 0 → ReLU output = 0

# Role of weight initialization
* Neural networks are sensitive to starting weights, especially with ReLU activations.
* Some initial weights:

  * Activate neurons well for all parts of the input → gradients flow → fast convergence → good predictions.

  * Activate neurons poorly for some inputs (dead ReLUs) → gradients are small or zero → slow convergence or stuck in a local minimum → bad predictions.


RandomUniform between -1 and 1 gives a wide range of starting points.

* Depending on how many neurons are “active” for early batches, Adam sees good or bad gradients:
  * Run 1 → mostly good initial activations → predictions reasonable
  * Run 3 → unlucky initialization → some neurons mostly dead → predictions terrible

# Related Concepts
* Dead neurons problem (ReLU): if pre-activation ≤ 0, output = 0, gradient = 0
* Local minima / saddle points: bad initializations can push optimization toward flat or suboptimal regions
* Variance of initialization matters: too high → exploding activations, too low → vanishing activations
